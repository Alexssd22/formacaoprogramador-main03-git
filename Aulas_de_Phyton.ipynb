```python
#muitas dificuldades em compreender esses links de perguntas, uma variável que assume um valor, depois outro, depois outro...



contas = []

def criar_conta():
    verificador_conta_existe = False
    dados_nova_conta = {}
    dados_nova_conta['numero_conta'] = input('Digite o número da nova conta:\n>>> ')

    for conta_em_consulta in contas:
        if conta_em_consulta['numero_conta'] == dados_nova_conta['numero_conta']:
            verificador_conta_existe = True
            break

    if verificador_conta_existe:
        print('\nO número da conta já existe. Tente novamente.')
    else:
        dados_nova_conta['saldo_conta'] = float(input('Digite o saldo da nova conta:\n>>> '))
        contas.append(dados_nova_conta)
        print('\nOperação efetuada com sucesso.')

def remover_conta():
    numero_conta = input('Digite o número da conta para remover:\n>>> ')
    conta_encontrada = False
    for conta_em_consulta in contas:
        if conta_em_consulta['numero_conta'] == numero_conta:
            conta_encontrada = True
            contas.remove(conta_em_consulta)
            print('\nOperação efetuada com sucesso.')
            break
    if not conta_encontrada:
        print('\nO número da conta não existe. Tente novamente.')

def listar_contas():
    if len(contas) == 0:
        print("\nNão há contas no sistema.\n")
    else:
        index = 1
        for conta_em_consulta in contas:
            if conta_em_consulta["saldo_conta"] >= 0:
                status = "Positivo"
            else:
                status = "Negativo"

            print(f"{index} - Número da Conta: {conta_em_consulta['numero_conta']} /// Saldo da conta R$: {conta_em_consulta['saldo_conta']:.2f} /// Status: {status}")
            index += 1

def adicionar_saldo_em_conta():
    numero_conta = input("Digite o número da conta: \n>>> ")
    conta_encontrada = False
    for conta_em_consulta in contas:
        if conta_em_consulta["numero_conta"] == numero_conta:
            conta_encontrada = True
            valor_credito = float(input("Digite o valor do depósito em R$: \n>>> "))
            if valor_credito < 0:
                print("\nValores negativos não são aceitos. Tente novamente.")
                return
            else:
                conta_em_consulta["saldo_conta"] += valor_credito
                print("\nSaldo atualizado com sucesso.")
            break
    if not conta_encontrada:
        print("\nO número da conta não existe. Tente novamente.")

def remover_saldo_em_conta():
    numero_conta = input("Digite o número da conta: \n>>> ")
    conta_encontrada = False
    for conta_em_consulta in contas:
        if conta_em_consulta["numero_conta"] == numero_conta:
            conta_encontrada = True
            valor_debito = float(input("Digite o valor de retirada em R$: \n>>> "))
            if valor_debito < 0:
                print("\nValores negativos não são aceitos. Tente novamente.")
                return
            elif valor_debito > conta_em_consulta["saldo_conta"]:
                print("\nSaldo insuficiente para essa retirada.")
                return
            else:
                conta_em_consulta["saldo_conta"] -= valor_debito
                print("\nSaldo atualizado com sucesso.")
            break
    if not conta_encontrada:
        print("\nO número da conta não existe. Tente novamente.")

def transferir_saldo():
    conta_partida = input("Digite o número da sua conta:\n>>> ")
    verificador_etapa = 0

    for conta_em_consulta in contas:
        if conta_em_consulta["numero_conta"] == conta_partida:
            verificador_etapa = 1
            conta_destino = input('Digite o número da conta destino:\n>>> ')

            if conta_destino != conta_partida:
                for conta_destino_em_consulta in contas:
                    if conta_destino_em_consulta['numero_conta'] == conta_destino:
                        verificador_etapa = 3
                        valor_transferencia = float(input('Digite o valor da transferência:\n>>> '))

                        if valor_transferencia < 0:
                            print('\nNão é possível transferir um valor negativo.')
                            break
                        elif valor_transferencia > conta_em_consulta['saldo_conta']:
                            print('\nNão é possível transferir um valor maior que o seu saldo.')
                            break
                        else:
                            conta_em_consulta['saldo_conta'] -= valor_transferencia
                            conta_destino_em_consulta['saldo_conta'] += valor_transferencia
                            print('\nOperação efetuada com sucesso.')
                            break
                break
            else:
                verificador_etapa = 2
                break

    if verificador_etapa == 0:
        print('\nA sua conta não foi encontrada. Tente novamente.')
    elif verificador_etapa == 1:
        print('\nA conta de destino não foi encontrada. Tente novamente.')
    elif verificador_etapa == 2:
        print('\nNão é possível transferir dinheiro para a mesma conta.')

def definir_saldo_em_conta():
    numero_conta = input("Digite o número da conta:\n>>> ")
    for conta_em_consulta in contas:
        # Comparação corrigida para numero_conta
        if conta_em_consulta["numero_conta"] == numero_conta:
            # Dois pontos e indentação corrigidos
            if conta_em_consulta["saldo_conta"] >= 0:
                status = "Positivo"
            else:
                status = "Negativo"

            # O print é executado para a conta localizada, positivo ou negativo
            print(f"\nO saldo em conta é: R$ {conta_em_consulta['saldo_conta']:.2f} - Status: {status}")
            return

    print("\nNúmero de conta não encontrado. Tente novamente.")


print(">>> Bem vindo ao sistema Bancário <<<")

while True:
    print(contas)
    print("\n### Menu ###")
    print("0 - Sair")
    print("1 - Criar uma nova conta")
    print("2 - Remover uma conta")
    print("3 - Listar contas existentes")
    print("4 - Adicionar Saldo")
    print("5 - Remover Saldo")
    print("6 - Transferir Saldo")
    print("7 - Saldo em conta")
    numero_operacao = input('Selecione a operação que deseja realizar:\n>>> ')

    if numero_operacao == "0":
        print('\nSistema encerrado.')
        break
    elif numero_operacao == "1":
        criar_conta()
    elif numero_operacao == "2":
        remover_conta()
    elif numero_operacao == "3":
        listar_contas()
    elif numero_operacao == "4":
        adicionar_saldo_em_conta()
    elif numero_operacao == "5":
        remover_saldo_em_conta()
    elif numero_operacao == "6":
        transferir_saldo()
    elif numero_operacao == "7":
        definir_saldo_em_conta()
    else:
        print('\nOperação inválida.')
```

```python
#Exercício 6: Mini-Estoque de Loja (Juntando Tudo)
#Objetivo: Crie do zero uma função chamada vender_produto(). Ela deve:
#Pedir ao usuário com input o nome do produto que ele quer comprar.
#Usar um for para percorrer a lista estoque abaixo.
#Se encontrar o produto, verificar se a "quantidade" dele é maior que 0.
#Se for maior que 0, diminuir 1 da "quantidade" e imprimir "Venda realizada!".
#Se for igual a 0, imprimir "Produto esgotado!".
#Se percorrer a lista inteira e o produto não existir, imprimir

atualizado = 0
estoque = [
    {"nome": "teclado", "quantidade": 5},
    {"nome": "mouse", "quantidade": 0},
    {"nome": "monitor", "quantidade": 2}
]

def vender_produto():
  produto_desejado = input("Digite o nome do produto que deseja comprar: ")
  produto_solicitado = False

  for produto in estoque:
    if produto["nome"] == produto_desejado:
      produto_solicitado = True
      if produto["quantidade"] > 0:
        produto["quantidade"] -= 1
        print("Produto vendido com sucesso")
        atualizado = 1

        if atualizado == 1:
           print("\n\nEstoque atualizado!\n\n")
           print(f"Produto: {produto['nome']}, Quantidade: {produto["quantidade"]}")
        break

      else:
        print("Produto esgotado")
        atualizado = 2
        break

    if not produto_solicitado:
      print("\nProduto não encontrado\n")
      atualizado = 0
      break


vender_produto()
```
